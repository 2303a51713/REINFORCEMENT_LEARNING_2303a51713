```python
import math

# ============================================================
# UPPER CONFIDENCE BOUND (UCB) - MULTI-ARMED BANDIT
# Reinforcement Learning - Practical 4
# ============================================================

# ------------------------------------------------------------
# PART A - MACHINE INITIALIZATION
# ------------------------------------------------------------

machines = ['A', 'B', 'C']

# Number of machines
M = 3

# Number of plays
N_play = 10

# ------------------------------------------------------------
# PART B - INITIALIZE KNOWLEDGE TABLES
# ------------------------------------------------------------

# N[a] = Number of times machine a has been selected
N = [0, 0, 0]

# R[a] = Total reward obtained from machine a
R = [0, 0, 0]

# Q[a] = Estimated value of machine a
Q = [0.0, 0.0, 0.0]

# UCB[a] = Upper Confidence Bound value
UCB = [float('inf')] * M

# Total reward received
total_reward = 0


# ------------------------------------------------------------
# FUNCTION TO CALCULATE UCB
# ------------------------------------------------------------

def calculate_ucb(total_plays):
    """
    UCB formula:

        UCB(a) = Q(a) + sqrt(2 * ln(t) / N(a))

    If a machine has never been selected, its UCB is infinity.
    This guarantees that every machine is tried at least once.
    """

    ucb_values = []

    for i in range(M):

        if N[i] == 0:
            # Unvisited machine
            ucb_values.append(float('inf'))

        else:
            # UCB calculation
            confidence = math.sqrt(
                (2 * math.log(total_plays)) / N[i]
            )

            ucb = Q[i] + confidence
            ucb_values.append(ucb)

    return ucb_values


# ------------------------------------------------------------
# FUNCTION TO DISPLAY CURRENT TABLE
# ------------------------------------------------------------

def display_table(ucb_values):

    print("\n" + "-" * 65)
    print(
        f"{'Machine':<12}"
        f"{'N (Count)':<15}"
        f"{'R (Reward)':<15}"
        f"{'UCB':<15}"
        f"{'Q-Value':<15}"
    )
    print("-" * 65)

    for i in range(M):

        if math.isinf(ucb_values[i]):
            ucb_display = "Infinity"
        else:
            ucb_display = f"{ucb_values[i]:.4f}"

        print(
            f"{machines[i]:<12}"
            f"{N[i]:<15}"
            f"{R[i]:<15}"
            f"{ucb_display:<15}"
            f"{Q[i]:<15.4f}"
        )

    print("-" * 65)


# ------------------------------------------------------------
# MAIN PROGRAM
# ------------------------------------------------------------

print("=" * 70)
print("     UPPER CONFIDENCE BOUND (UCB) - MULTI-ARMED BANDIT")
print("=" * 70)

print("\nMachines available:", machines)
print("Number of plays:", N_play)

print("\nInitial Values:")
print("N =", N)
print("R =", R)
print("Q =", Q)


# ------------------------------------------------------------
# PART C, D AND E
# UCB CALCULATION, REWARD COLLECTION AND Q UPDATE
# ------------------------------------------------------------

for play in range(1, N_play + 1):

    print("\n" + "=" * 70)
    print("PLAY", play)
    print("=" * 70)

    # --------------------------------------------------------
    # Calculate UCB
    # --------------------------------------------------------

    # total number of plays before selecting current machine
    total_plays = play

    UCB = calculate_ucb(total_plays)

    print("\nCurrent UCB values:")

    for i in range(M):

        if math.isinf(UCB[i]):
            print(f"Machine {machines[i]} : Infinity")
        else:
            print(f"Machine {machines[i]} : {UCB[i]:.4f}")

    # --------------------------------------------------------
    # Select machine with highest UCB
    # --------------------------------------------------------

    selected_index = UCB.index(max(UCB))
    selected_machine = machines[selected_index]

    print("\nSelected Machine:", selected_machine)

    # --------------------------------------------------------
    # PART D - GET REWARD FROM USER
    # --------------------------------------------------------

    while True:

        try:

            reward = float(
                input(
                    f"Enter reward obtained from Machine "
                    f"{selected_machine}: "
                )
            )

            break

        except ValueError:

            print("Please enter a valid numerical reward.")

    # --------------------------------------------------------
    # Update total reward
    # --------------------------------------------------------

    total_reward += reward

    # --------------------------------------------------------
    # Update N
    # --------------------------------------------------------

    N[selected_index] += 1

    # --------------------------------------------------------
    # Update R
    # --------------------------------------------------------

    R[selected_index] += reward

    # --------------------------------------------------------
    # PART E - INCREMENTAL SAMPLE-AVERAGE Q UPDATE
    # --------------------------------------------------------

    old_Q = Q[selected_index]

    Q[selected_index] = (
        old_Q
        + (1 / N[selected_index])
        * (reward - old_Q)
    )

    # --------------------------------------------------------
    # Display update
    # --------------------------------------------------------

    print("\nUpdated values for Machine", selected_machine)

    print("Previous Q-value :", round(old_Q, 4))
    print("Reward          :", reward)
    print("Selection count :", N[selected_index])
    print("New Q-value     :", round(Q[selected_index], 4))

    # --------------------------------------------------------
    # Show current tables
    # --------------------------------------------------------

    UCB = calculate_ucb(play)

    print("\nCurrent Table:")
    display_table(UCB)


# ============================================================
# PART F - FINAL RESULTS
# ============================================================

print("\n\n")
print("=" * 70)
print("                    FINAL RESULTS")
print("=" * 70)

# Calculate final UCB values
UCB = calculate_ucb(N_play)

print("\nFinal N, R, UCB and Q Tables:")

display_table(UCB)


# ------------------------------------------------------------
# TOTAL REWARD
# ------------------------------------------------------------

print("\nTotal Reward Received =", total_reward)


# ------------------------------------------------------------
# AVERAGE REWARD
# ------------------------------------------------------------

average_reward = total_reward / N_play

print("Average Reward        =", round(average_reward, 4))


# ------------------------------------------------------------
# BEST MACHINE
# ------------------------------------------------------------

best_machine_index = Q.index(max(Q))
best_machine = machines[best_machine_index]

print("\nBest Machine based on Q-value:", best_machine)
print("Highest Q-value:", round(Q[best_machine_index], 4))


# ------------------------------------------------------------
# FINAL ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("                         ANALYSIS")
print("=" * 70)

for i in range(M):

    print(
        f"Machine {machines[i]} was selected "
        f"{N[i]} time(s), "
        f"received total reward {R[i]}, "
        f"and has estimated Q-value {Q[i]:.4f}."
    )

print("\nThe machine with the highest Q-value is Machine", best_machine)
print("This indicates that the random agent estimates Machine",
      best_machine, "to provide the highest average reward.")

print("\nProgram completed successfully.")
print("=" * 70)
```
