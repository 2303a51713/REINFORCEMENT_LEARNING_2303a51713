```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Parameters
N_play = 10
epsilon = 0.2

# Initialize N, R and Q tables
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Highest Q-value machine
    best_machine = max(Q, key=Q.get)

    # Modified epsilon-greedy decision
    if r < epsilon:
        # Exploration: exclude highest-Q machine
        other_machines = [m for m in machines if m != best_machine]
        selected = random.choice(other_machines)
        decision = "Exploration"
    else:
        # Exploitation
        selected = best_machine
        decision = "Exploitation"

    print(f"\nPlay {play}")
    print(f"Decision: {decision}")
    print(f"Play with machine {selected}")

    # Reward collection
    reward = float(input("Enter reward: "))

    # Update N
    N[selected] += 1

    # Update R
    R[selected] += reward

    # Incremental sample-average update
    Q[selected] = Q[selected] + (
        (reward - Q[selected]) / N[selected]
    )

    print(f"Reward = {reward}")
    print(f"N[{selected}] = {N[selected]}")
    print(f"Q[{selected}] = {Q[selected]:.2f}")


# Part F - Display Results
print("\n========== FINAL RESULTS ==========")

print("\nN Table:")
for machine in machines:
    print(machine, "=", N[machine])

print("\nR Table:")
for machine in machines:
    print(machine, "=", R[machine])

print("\nQ Table:")
for machine in machines:
    print(machine, "=", round(Q[machine], 2))

# Total reward
total_reward = sum(R.values())

print("\nTotal Reward =", total_reward)
```
