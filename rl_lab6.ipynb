```python
import gymnasium as gym

def run_random_agent(env_name, num_episodes=5):
    print(f"\n{'='*40}")
    print(f"Environment: {env_name}")
    print(f"{'='*40}")

    # 1. Create the selected environment
    env = gym.make(env_name)

    for episode in range(1, num_episodes + 1):
        # 2. Reset the environment
        # 3. Observe the initial state
        state, info = env.reset()

        total_reward = 0
        steps = 0
        done = False

        # 8. Continue until the episode terminates
        while not done:
            # 4. Identify the available actions (handled by action_space)
            # 5. Select an action randomly
            action = env.action_space.sample()

            # 6. Execute the action in the environment
            # 7. Observe the next state and reward
            next_state, reward, terminated, truncated, info = env.step(action)

            total_reward += reward
            steps += 1

            # Update state and check if the episode is finished
            state = next_state
            done = terminated or truncated

        # 9. Calculate the average reward obtained during the episode
        average_reward = total_reward / steps if steps > 0 else 0

        # 10. Repeat and display total and average reward for each episode
        print(f"Episode {episode}: "
              f"Steps = {steps:<5} | "
              f"Total Reward = {total_reward:<5} | "
              f"Average Reward = {average_reward:.2f}")

    env.close()

if __name__ == "__main__":
    # Test on CliffWalking
    run_random_agent("CliffWalking-v1", num_episodes=5)

    # Test on Taxi (Texi)
    run_random_agent("Taxi-v4", num_episodes=5)
```
