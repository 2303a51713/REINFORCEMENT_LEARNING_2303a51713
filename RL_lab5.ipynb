```python
import gymnasium as gym


def random_agent(env_name):

    # Create environment
    if env_name == "CartPole":
        env = gym.make("CartPole-v1")

    elif env_name == "MountainCar":
        env = gym.make("MountainCar-v0")

    elif env_name == "FrozenLake":
        env = gym.make("FrozenLake-v1", is_slippery=True)

    else:
        print("Invalid environment")
        return

    # Reset environment
    state, info = env.reset()

    total_reward = 0
    steps = 0

    print("\nEnvironment:", env_name)
    print("Initial State:", state)
    print("Action Space:", env.action_space)

    while True:

        # Select random action
        action = env.action_space.sample()

        # Execute action
        next_state, reward, terminated, truncated, info = env.step(action)

        print(
            "Step:", steps + 1,
            "Action:", action,
            "Reward:", reward,
            "Next State:", next_state
        )

        total_reward += reward
        steps += 1

        state = next_state

        if terminated or truncated:
            break

    # Calculate average reward
    average_reward = total_reward / steps

    print("\nEpisode Completed")
    print("Total Steps:", steps)
    print("Total Reward:", total_reward)
    print("Average Reward:", round(average_reward, 4))

    env.close()


# -------------------------------------------------
# MAIN
# -------------------------------------------------

print("RANDOM AGENT - REINFORCEMENT LEARNING")

print("\n1. CartPole")
print("2. MountainCar")
print("3. FrozenLake")

choice = input("\nEnter choice: ")

if choice == "1":
    random_agent("CartPole")

elif choice == "2":
    random_agent("MountainCar")

elif choice == "3":
    random_agent("FrozenLake")

else:
    print("Invalid choice")
```
